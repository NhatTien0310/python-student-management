```python
 #1. Tạo DataFrame

import pandas as pd

data = {
    "Ho_ten": [
        "Nguyen Van An",
        "Tran Thi Binh",
        "Le Van Cuong",
        "Pham Thi Dung",
        "Hoang Van Em",
        "Vo Thi Giang",
        "Dang Van Hung",
        "Bui Thi Lan",
        "Do Van Minh",
        "Nguyen Thi Nga"
    ],
    "Chuyen_can": [9, 8, 7, 10, 6, 9, 8, 7, 5, 10],
    "Giua_ky": [8, 7, 6, 9, 5, 8, 7, 6, 4, 9],
    "Cuoi_ky": [9, 8, 7, 9, 6, 9, 8, 7, 5, 10]
}

df = pd.DataFrame(data)

display(df)
```

```python
 #2. Tính điểm tổng kết
df["Tong_ket"] = (
    0.2 * df["Chuyen_can"]
    + 0.3 * df["Giua_ky"]
    + 0.5 * df["Cuoi_ky"]
)
display(df)
```

```python
 #3. Xep loai sinh vien

def xep_loai(diem):
    if diem >= 8.5:
        return "Gioi"
    elif diem >= 7.0:
        return "Kha"
    elif diem >= 5.0:
        return "Trung binh"
    else:
        return "Yeu"

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)

display(df)
```

```python
#4. Thong ke ket qua hoc tap

# 1. Tinh diem trung binh cua lop
diem_tb = df["Tong_ket"].mean()

# 2. Tim sinh vien co diem cao nhat
sinh_vien_cao_nhat = df.loc[df["Tong_ket"].idxmax()]

# 3. Tim sinh vien co diem thap nhat
sinh_vien_thap_nhat = df.loc[df["Tong_ket"].idxmin()]

# 4. Dem so sinh vien dat tu 5 diem tro len
so_sinh_vien_dat = (df["Tong_ket"] >= 5).sum()

# Hien thi ket qua
print("Diem trung binh cua lop:", round(diem_tb, 2))

print("\nSinh vien co diem cao nhat:")
print(sinh_vien_cao_nhat)

print("\nSinh vien co diem thap nhat:")
print(sinh_vien_thap_nhat)

print("\nSo sinh vien dat:", so_sinh_vien_dat)
```

```python
 #5. Ve bieu do cot diem tong ket

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(df["Ho_ten"], df["Tong_ket"])

plt.title("DIEM TONG KET CUA 10 SINH VIEN")
plt.xlabel("Ho ten sinh vien")
plt.ylabel("Diem tong ket")

plt.ylim(0, 10)
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
